# pose estimation

```python
def prediction(model_name, img_path, display_result=False, task: str =None):
    model = YOLO(model_name)
    result = model(img_path, save=True, conf=0.7)

    img_filename = img_path.split("/")[-1].split(".")[0]

    for r in result: 
        pred_img_path = f'{r.save_dir}/{img_filename}.jpg'
        pred = cv2.cvtColor(cv2.imread(pred_img_path), cv2.COLOR_BGR2RGB)
        plt.imshow(pred)
        plt.axis('off')
        plt.title(f"YOLO - {task}")
    plt.show()

    if display_result: print(result)

    return result
 
```

```python
result = prediction("yolo11l-pose.pt", "data/fb.png")
```
